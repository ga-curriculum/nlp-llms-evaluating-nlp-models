# Intro to Hugging Face Evaluate   
Let's start by loading our metrics from the Hugging Face library using:
```python 
evaluate.load('metric-of-choice')
``` 
For example, you could load accuracy or F1-Score with:

```python 
# Load standard metics 
accuracy_metric = evaluate.load('accuracy')
f1_metric = evaluate.load('f1')
```

Here you will load **ROUGE** to evaluate summaries and **SacreBLEU** to evaluate translations.

````python
# load specialized evaluation metrics
rouge = evaluate.load('rouge')
bleu = evaluate.load('sacrebleu')
````

# Evaluate Summaries with ROUGE Scores
Hugging Face evaluate outputs **ROUGE-1**, **ROUGE-2**, and **ROUGE-L** scores.

We call the metrics with:

```python
rouge = evaluate.load('rouge')
rouge.compute(predictions=[model_summary],
                           references=[human_summary],
                           use_aggregator=True)
```
This returns a dictionary with keys for `rouge1`, `rouge2`, `rougeL`, and `rougeLsum`. You can choose to output all them or review a subset for your analysis.

# Try It! 
Calculate the **ROUGE** Scores comparing that model summary to the human summary for our reviews. How did the model perform?

````python
# calculate ROUGE scores
rouge_scores = rouge.compute(predictions=[model_summary],
                           references=[human_summary],
                           use_aggregator=True)


rouge_scores
````

````python
# print only the scores you need
print("Selected Rouge Scores")
print("ROUGE-1:", rouge_scores['rouge1'])
print("ROUGE-2:", rouge_scores['rouge2'])
print("ROUGE-L:", rouge_scores['rougeL'])
````

# Evaluate Translations with SacreBLEU
You can load any specialized (or common!) metric in a similar fashion.

Below is the syntax for **SacreBLEU**:

```python
#load sacreBLEU metric
bleu = evaluate.load('sacrebleu')

# Calculate BLEU score
bleu_score = bleu.compute(predictions=[model_translation],
                         references=[[reference_translation]])
```
This outputs a dictionary with the following keys: `score`, `counts`, `totals`, `precisions`, `bp`, `sys_len`, `ref_len`

The following code will output only the **SacreBLEU** score:

```python
print(bleu_score['score'])
```
